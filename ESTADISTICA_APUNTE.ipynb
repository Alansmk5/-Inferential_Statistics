```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Crear una población
poblacion = np.random.uniform(0, 100, 10000)

# Graficar la población original
plt.hist(poblacion, bins=30, edgecolor='black')
plt.title("Distribución de la población")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.show()


# 2. Tomar 1000 muestras de tamaño 30
# y calcular la media de cada muestra
medias = []

for i in range(1000):
    muestra = np.random.choice(poblacion, size=30)
    medias.append(np.mean(muestra))

# Graficar las medias muestrales
plt.hist(medias, bins=30, edgecolor='black')
plt.title("Distribución de las medias muestrales")
plt.xlabel("Media")
plt.ylabel("Frecuencia")
plt.show()


# 3. Comparar diferentes tamaños de muestra
tamanos = [5, 30, 100]

for n in tamanos:
    medias = []

    for i in range(1000):
        muestra = np.random.choice(poblacion, size=n)
        medias.append(np.mean(muestra))

    plt.hist(medias, bins=30, edgecolor='black')
    plt.title(f"Medias muestrales con n = {n}")
    plt.xlabel("Media")
    plt.ylabel("Frecuencia")
    plt.show()
```
